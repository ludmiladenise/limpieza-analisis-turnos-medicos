**3. LIMPIEZA**

```python
# Crear una copia para conservar el dataset original
df_limpio = df.copy()

# Revisar valores nulos
print("Valores nulos por columna:")
print(df_limpio.isnull().sum())

# Revisar registros duplicados
print("Cantidad de duplicados:", df_limpio.duplicated().sum())

# Revisar edades negativas
print("Edades negativas:", (df_limpio['Age'] < 0).sum())
```

```python
# Crear una copia del dataset original
df_limpio = df.copy()

# Contar los registros con edad negativa
print("Registros con edad negativa:", (df_limpio['Age'] < 0).sum())

# Eliminar registros con edad negativa
df_limpio = df_limpio[df_limpio['Age'] >= 0].copy()

# Reiniciar el índice
df_limpio = df_limpio.reset_index(drop=True)

# Verificar el resultado
print("Dimensiones después de la limpieza:", df_limpio.shape)
print("Edades negativas restantes:", (df_limpio['Age'] < 0).sum())
```

```python
#Verificar y convertir las columnas de fechas.

# Convertir las columnas de fecha
df_limpio['ScheduledDay'] = pd.to_datetime(
    df_limpio['ScheduledDay'], errors='coerce'
)

df_limpio['AppointmentDay'] = pd.to_datetime(
    df_limpio['AppointmentDay'], errors='coerce'
)

# Comprobar los tipos de datos
print(df_limpio[['ScheduledDay', 'AppointmentDay']].dtypes)
```

```python
# Verificar si quedaron fechas faltantes después de la conversión
print("Fechas faltantes en ScheduledDay:",
      df_limpio['ScheduledDay'].isnull().sum())

print("Fechas faltantes en AppointmentDay:",
      df_limpio['AppointmentDay'].isnull().sum())
```

**4. ANÁLISIS.**

```python
# Cantidad de turnos según asistencia
resultado_turnos = df_limpio.groupby('No-show').size()

print(resultado_turnos)
```

```python
porcentaje_turnos = (
    df_limpio['No-show'].value_counts(normalize=True) * 100
)

print(porcentaje_turnos.round(2))
```

### Análisis de asistencia a los turnos médicos

Luego de realizar la limpieza del dataset, se analizaron 110.526 turnos médicos. Los resultados muestran que 88.207 pacientes asistieron a sus turnos, lo que representa el 79,81% del total. Por otro lado, 22.319 turnos correspondieron a inasistencias, equivalentes al 20,19%.

Estos resultados indican que aproximadamente 2 de cada 10 turnos programados no registraron asistencia. Esta información puede ser útil para estudiar los factores asociados a las inasistencias y proponer estrategias para mejorar la asistencia de los pacientes.

Ahora vamos a analizar si existe una diferencia en las inasistencias según la recepción de SMS.

```python
inasistencias_sms = df_limpio.groupby('SMS_received').agg(
    Total_turnos=('No-show', 'count'),
    Inasistencias=('No-show', lambda x: (x == 'Yes').sum())
)

inasistencias_sms['Porcentaje_inasistencia'] = (
    inasistencias_sms['Inasistencias']
    / inasistencias_sms['Total_turnos'] * 100
)

display(inasistencias_sms.round(2))
```

### Análisis de inasistencias según recepción de SMS

Se agrupó el dataset según la variable SMS_received para comparar la tasa de inasistencia entre los pacientes que recibieron un mensaje y aquellos que no lo recibieron.

Se observó que, entre los 75.044 turnos sin SMS registrado, 12.535 correspondieron a inasistencias, lo que representa el 16,70%. En cambio, de los 35.482 turnos con SMS registrado, 9.784 fueron inasistencias, equivalentes al 27,57%.

La diferencia observada fue de 10,87 puntos porcentuales. Este resultado permite identificar una asociación entre la recepción de SMS y la inasistencia en el conjunto analizado, pero no demuestra una relación causal.

Ahora vamos a identificar los barrios con mayor cantidad de turnos en los que los pacientes no asistieron.

```python
inasistencias_barrio = df_limpio.groupby('Neighbourhood').agg(
    Total_turnos=('No-show', 'count'),
    Inasistencias=('No-show', lambda x: (x == 'Yes').sum())
)

inasistencias_barrio['Porcentaje_inasistencia'] = (
    inasistencias_barrio['Inasistencias']
    / inasistencias_barrio['Total_turnos'] * 100
)

inasistencias_barrio = inasistencias_barrio.sort_values(
    by='Inasistencias',
    ascending=False
)

display(inasistencias_barrio.head(10).round(2))
```

### Análisis de inasistencias por barrio

Se agruparon los turnos por barrio utilizando la variable Neighbourhood, calculando el total de turnos, la cantidad de inasistencias y el porcentaje de inasistencia.

Entre los diez barrios con mayor cantidad absoluta de inasistencias, Jardim Camburi registró 1.465 casos, con una tasa del 18,98%. Por otro lado, Itararé presentó el porcentaje de inasistencia más elevado de este grupo, con un 26,27%, mientras que Jardim da Penha registró el menor porcentaje, con un 16,28%.

Estos resultados permiten observar diferencias en el comportamiento de las inasistencias entre los barrios. Sin embargo, para interpretar adecuadamente estas diferencias, es necesario considerar también el volumen total de turnos de cada zona.
